# Carga de datos: Telco Customer Churn

Este cuaderno carga el dataset del proyecto y comprueba que se leyó bien.

- **Fuente:** Kaggle, Telco Customer Churn (BlastChar)
- **Enlace:** https://www.kaggle.com/datasets/blastchar/telco-customer-churn
- **Fecha de obtención:** 22/10/2026
- **Archivo:** `data/WA_Fn-UseC_-Telco-Customer-Churn.csv`

Se ejecuta de inicio a fin en Google Colab, sin pasos manuales.

```python
# 1. Preparación del entorno
import os
import random
import hashlib
import subprocess
import numpy as np
import pandas as pd

# Semilla fija para que los resultados sean siempre los mismos
SEMILLA = 42
random.seed(SEMILLA)
np.random.seed(SEMILLA)

# Enlace del repositorio de GitHub del grupo
REPO_URL = "https://github.com/sofiaenma14/churn-telecom-proyecto-integrador.git"
CARPETA_REPO = REPO_URL.rstrip("/").split("/")[-1].replace(".git", "")

# Saber si se está ejecutando en Google Colab
try:
    import google.colab
    EN_COLAB = True
except ImportError:
    EN_COLAB = False

# En Colab se descarga el repositorio y se entra a su carpeta
if EN_COLAB and not os.path.exists("data"):
    if "USUARIO" in REPO_URL:
        raise ValueError("Falta pegar el enlace real del repositorio en REPO_URL.")
    if not os.path.exists(CARPETA_REPO):
        subprocess.run(["git", "clone", REPO_URL], check=True)
    os.chdir(CARPETA_REPO)
```

```python
# 2. Carga del archivo con una ruta relativa a la carpeta del repositorio
RUTA_CSV = "data/WA_Fn-UseC_-Telco-Customer-Churn.csv"

# Si el cuaderno se abre desde la carpeta notebooks/, se sube un nivel
if not os.path.exists(RUTA_CSV):
    RUTA_CSV = "../" + RUTA_CSV

df = pd.read_csv(RUTA_CSV)
print("Datos cargados correctamente")
```

```python
# 3. Comprobaciones de que el archivo es el correcto

# Código de verificación: confirma que el archivo no cambió
SHA256_ESPERADO = "88be4b93fbe0cc83421af1c503794c97c342eca914c1576db7c276e61d61358a"
with open(RUTA_CSV, "rb") as f:
    sha256_actual = hashlib.sha256(f.read()).hexdigest()
print("Código de verificación coincide:", sha256_actual == SHA256_ESPERADO)

# Tamaño esperado: 7 043 filas y 21 columnas
print("Filas y columnas:", df.shape)

# Cuántos clientes se dieron de baja (variable objetivo Churn)
print(df["Churn"].value_counts())

# Línea base: porcentaje de clientes que se dieron de baja
linea_base = (df["Churn"] == "Yes").mean() * 100
print("Línea base (tasa de fuga):", round(linea_base, 2))
```
